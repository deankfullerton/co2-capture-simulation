```python
# ASPEN SIMULATION RESULTS

# Stream Results
co2_captured_tonne_hr = 152.75985905   # tonne CO2/hr, from stream results converted from kg/sec
mea_makeup_rate = 0.184051217205544 * 60 * 60 # kg/hr

# Equipment Results
reboiler_duty_GJ_hr = 641.1636         # GJ/hr (178.101 MW converted)
reboiler_duty_GJ_tonne = reboiler_duty_GJ_hr/co2_captured_tonne_hr  # GJ/tonne CO2, calculated from above
```

```python
# TEA ASSUMPTIONS
plant_operating_hours = 8000         # hr/yr, standard industry assumption
co2_annual_tonne = co2_captured_tonne_hr * plant_operating_hours
mea_price_per_kg = 1.04 # $/kg; Sourced Online

# Steam Price
steam_cost_per_hr = 7427.148454          # USD/hr, from ASPEN utilities sheet
steam_mass = 912.4261 * 1000 * (1/2.205) # kg/hr; converted from klb/hr
energy_conversion = 2750 # kJ/kg
GJ_hr = (steam_mass * energy_conversion)/1_000_000  #GJ/hr
ASPEN_steam_price = steam_cost_per_hr / GJ_hr

plant_capacity_tonne_day = (co2_captured_tonne_hr * plant_operating_hours) / 365 #tonne CO₂/day
plant_lifetime = 20 #years
discount_rate = 0.10 #10%

print(f"TEA ASSUMPTIONS")
print(f"Steam Price: ${ASPEN_steam_price:,.2f}/GJ")
print(f"Price of MEA: ${mea_price_per_kg:,.2f}/kg")
print(f"Plant Capacity: {co2_annual_tonne} tonne CO2/year")
```

```python
# ASPEN ECONOMIC RESULTS

# Installed costs from ASPEN Economic Analyzer (Icarus)
# Units: USD
cost_absorber = 6508000
cost_stripper = 3925000
cost_cooler = 1906200
cost_heater = 335800
cost_flash = 431300
cost_pumps = 368400 + 46700
TIC = 13521400

# Utility costs from ASPEN Economic Analyzer
steam_cost_per_hr = 7427.148454          # USD/hr, from ASPEN utilities sheet
cooling_cost_per_hr = 371.5236         # USD/hr
electricity_cost_per_hr = 9.923022       # USD/hr

print(f"INSTALLATION COSTS")
print(f"Absorber: ${cost_absorber:,.0f}")
print(f"Stripper: ${cost_stripper:,.0f}")
print(f"Cooler: ${cost_cooler:,.0f}")
print(f"Heater: ${cost_heater:,.0f}")
print(f"Flash: ${cost_flash:,.0f}")
print(f"Pumps: ${cost_pumps:,.0f}")
print(f"Total Installation Cost: ${TIC:,.0f}")

print(f"\nUTILITIES COST")
print(f"Steam: ${steam_cost_per_hr:,.0f}/hr")
print(f"Cooling: ${cooling_cost_per_hr:,.0f}/hr")
print(f"Electricity: ${electricity_cost_per_hr:,.0f}/hr")
```

```python
# SOLVENT MAKEUP COST
mea_cost_per_hr = mea_makeup_rate * mea_price_per_kg
print(f"MEA Cost: ${mea_cost_per_hr:,.0f}/hr")
```

```python
# ANNUAL OPERATING COSTS
# Calculated at 8,000 operating hours per year
annual_mea_cost = mea_cost_per_hr * plant_operating_hours
annual_steam = steam_cost_per_hr * plant_operating_hours
annual_cooling = cooling_cost_per_hr * plant_operating_hours
annual_electricity = electricity_cost_per_hr * plant_operating_hours
annual_opex = annual_steam + annual_cooling + annual_electricity + annual_mea_cost

print(f"Total annual operating cost: ${annual_opex:,.0f}")
```

```python
# Capital Recovery Factor (CRF)
# Converts total installed cost into equivalent annual capital cost
# Capital recovery factor
CRF = (discount_rate * (1+discount_rate)**plant_lifetime) / ((1+discount_rate)**plant_lifetime - 1)
print(f"Capital Recovery Factor: {CRF:.4f}")
```

```python
# Annualized capital cost
annualized_capex = TIC * CRF

# Levelized cost
LCOC = (annualized_capex + annual_opex) / co2_annual_tonne

print(f"Annualized capital cost: ${annualized_capex:,.0f}/yr")
print(f"Annual CO2 captured: {co2_annual_tonne:,.0f} tonne/yr")
print(f"\nLevelized cost of capture: ${LCOC:.2f}/tonne CO2")
```
